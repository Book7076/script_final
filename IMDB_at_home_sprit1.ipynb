### 1. Project Title

**IMDB at home** (Movie Sentiment & Watchlist Manager)

### 2. Problem Statement

ผู้ใช้งานทั่วไปมักประสบปัญหาความยุ่งยากในการค้นหาภาพยนตร์ที่ตรงกับอารมณ์ความรู้สึกในขณะนั้น และขาดระบบจัดเก็บรายการภาพยนตร์ส่วนตัวที่บันทึกบทวิเคราะห์หรือความรู้สึกหลังรับชมได้อย่างเป็นระบบ

### 3. Proposed Solution

**IMDB at home** เป็น Full-Stack Web Application ที่ช่วยให้ผู้ใช้ค้นหาข้อมูลภาพยนตร์ ประเมินบรรยากาศของภาพยนตร์ผ่านระบบ Sentiment Analysis จากบทวิเคราะห์/เรื่องย่อ และจัดเก็บคลังภาพยนตร์ส่วนตัว (Watchlist) พร้อมระบบบันทึกรีวิว การให้คะแนน และสรุปสถิติการรับชมส่วนบุคคล

### 4. Domain

✅ Data Analysis & Management / Daily Apps

*(❌ Not Allowed: Information Systems, Financial Apps, Government Apps, ERP)*

### 5. API(s) to Use

* **API Name:** TMDB (The Movie Database) API


* **Documentation Link:** [https://developer.themoviedb.org/docs/getting-started](https://developer.themoviedb.org/docs/getting-started)
* **Type of Data:** Movie Metadata (Title, Overview, Genres, Rating, Release Date, Poster Path) ในรูปแบบ JSON



### 6. Data Persistence Plan & Layer Separation

* **Data Access Layer:** Mini database (SQLite) สำหรับตาราง `watchlist` พร้อมระบบ File I/O (Export เป็น CSV/JSON)


* **Business Logic Layer:** Python FastAPI สำหรับประมวลผล Sentiment, อัลกอริทึมการค้นหา (Searching), กรองข้อมูล (Filtering) และเรียงลำดับ (Sorting)


* **Presentation Layer:** React.js + Tailwind CSS และระบบตรวจความถูกต้องของอินพุต (Input Validation)

### รายชื่อ Use Cases ทั้งหมด (12 Use Cases)

#### 🎯 ส่วน MVP (10 Core Functions)

1. **UC-01:** Search Movies by Title (ยิงค้นหาผ่าน TMDB API)


2. **UC-02:** Fetch Trending & Top-Rated Movies (ดึงรายการภาพยนตร์ยอดนิยม)


3. **UC-03:** View Detailed Movie Profiles (แสดงรายละเอียดเชิงลึกของภาพยนตร์)


4. **UC-04:** Analyze Overview Sentiment (วิเคราะห์อารมณ์ของเรื่องย่อด้วย NLP)


5. **UC-05:** Add Movie to Watchlist (บันทึกภาพยนตร์ลงคลังส่วนตัว)


6. **UC-06:** View, Filter & Sort Watchlist (เรียกดู กรอง และเรียงลำดับรายการตามสถานะ)


7. **UC-07:** Update Watchlist Status & Rating (อัปเดตสถานะ ให้คะแนน และรีวิวส่วนตัว)


8. **UC-08:** Remove Movie from Watchlist (ลบรายการภาพยนตร์ออกจากคลัง)


9. **UC-09:** Export Watchlist to CSV/JSON (ส่งออกข้อมูลคลังภาพยนตร์เป็นไฟล์)


10. **UC-10:** Generate Personal Statistics (สรุปสถิติการดูหนังและประเภทโปรด)



#### 🚀 ส่วน Stretch Features (2 Optional Functions)

11. **UC-11:** Smart Mood Recommendation (ระบบแนะนำภาพยนตร์ตามอารมณ์ที่ระบุ)
12. **UC-12:** Backup & Restore System (ระบบสำรองและฟื้นฟูข้อมูลฐานข้อมูล SQLite)
