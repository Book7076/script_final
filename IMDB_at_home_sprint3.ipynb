### 2. โจทย์ เป้าหมาย และเงื่อนไขความเสร็จสมบูรณ์ (Objectives & Definition of Done)

* **โจทย์ & เป้าหมายประจำ Sprint 3:** เชื่อมต่อระบบ Front-End และ Back-End เข้าด้วยกันอย่างสมบูรณ์แบบ (Full-Stack Integration), จัดการสถานะแอปพลิเคชัน (State Management) และความสอดคล้องของข้อมูล (Data Consistency), พร้อมเสริมความทนทานต่อกรณีขอบเขต (Edge Cases Resilience)
* **Definition of Done (DoD):**
  1. **Full-Stack Wiring:** UI สามารถเรียกใช้งานฟังก์ชัน CRUD, Search, Filter, Sort และ Export ของ Back-End ได้อย่างไร้รอยต่อ
  2. **Data Consistency & State:** ข้อมูลในหน่วยความจำ (RAM/State) และไฟล์บันทึกข้อมูล (SQLite/JSON) ต้องสอดคล้องกันตลอดเวลาเมื่อมีการอัปเดตสถานะ
  3. **Edge Case Resilience:** ระบบต้องมี try-except ครอบคลุมทุกชั้น สามารถจัดการ Error (เช่น ไฟล์ DB เสียหาย, ค่าอินพุตติดลบ, ค้นหาค่าว่าง) โดยแสดงแจ้งเตือนอย่างเหมาะสมและโปรแกรมไม่ Crash 100%

### 3. สรุปความก้าวหน้าและการพัฒนาจริง (Execution & Code Details)

* [x] **System Integration Wiring:** เชื่อมต่อ UI Components (React/CLI) เข้ากับ FastAPI Endpoints ผ่าน Service Layer (`api.js`) รองรับ CRUD เต็มรูปแบบ
* [x] **State Management & Persistence:** จัดการ React State / Application State ให้ซิงค์ตรงกับฐานข้อมูล SQLite และไฟล์ JSON เมื่อผู้ใช้กดเพิ่ม/ลบ/แก้ไข Watchlist
* [x] **Search, Filter & Sort Integration:** เชื่อมต่ออัลกอริทึมการค้นหาชื่อหนัง, การกรองตาม Genre/Status และการเรียงลำดับคะแนน บน UI เข้ากับ BLL
* [x] **Edge Case & Exception Resilience Engine:**
  * เพิ่มระบบ **Auto-Recovery:** หากไม่พบไฟล์ DB/JSON ระบบจะสร้างไฟล์และ Schema เริ่มต้นให้อัตโนมัติทันที
  * ป้องกัน **Boundary Values:** ดักจับการกรอกคะแนนรีวิวติดลบ หรือค่าที่เกินขอบเขต 0–10 ด้วย Custom Exception
  * ป้องกัน **Empty Query Search:** ตัดช่องว่างด้วย `.strip()` และดักจับคำค้นหาที่เป็นค่าว่าง ไม่ให้ส่ง Request ที่ผิดพลาด
